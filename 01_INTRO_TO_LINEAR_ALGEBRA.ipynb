```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 40, 500) # start, finish, n points
t
```

```python
distance_robber = 2.5 * t
distance_sheriff = 3 *(t-5) # started 5 min late
```

```python
fig, ax = plt.subplots()
plt.title("A bank robber Caught")
plt.xlabel("time(in minutes)")
plt.ylabel("distance_in_km")
ax.set_xlim([0, 40])
ax.set_ylim([0,100])
ax.plot(t,distance_robber,c= "green")
ax.plot(t,distance_sheriff,c= "blue")
plt.axvline(x=30, color='red', linestyle='--')
plt.axhline(y=75, color='red', linestyle='--')


```
