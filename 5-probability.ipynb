## What Probability Theory Is

### Events and Sample Spaces

Let's assume we have a fair coin, which is equally likely to come up heads (H) or tails (T).

In instances like this, where the two outcomes are equally likely, we can use probability theory to express the likelihood of a particular event by comparing it with the sample space (the set of all possible outcomes; can be denoted as $\Omega$):

\[
P(\text{event}) = \frac{\# \text{ of outcomes of event}}{\# \text{ of outcomes in } \Omega}
\]

If we're only flipping the coin once, then there are only two possible outcomes in the sample space $\Omega$: it will either be H or T (using set notation, we could write this as $\Omega = \{H, T\}$).

Therefore:

\[
P(H) = \frac{1}{2} = 0.5
\]

Equally:

\[
P(T) = \frac{1}{2} = 0.5
\]

As a separate example, consider drawing a single card from a standard deck of 52 playing cards. In this case, the number of possible outcomes in the sample space $\Omega$ is 52.

There is only one ace of spades in the deck, so the probability of drawing it is:

\[
P(\text{ace of spades}) = \frac{1}{52} \approx 0.019
\]

In contrast there are four aces, so the probability of drawing an ace is:

\[
P(\text{ace}) = \frac{4}{52} = \frac{1}{13} \approx 0.077
\]

Some additional examples:

\[
P(\text{heads}) = \frac{1}{2} = 0.5
\]

\[
P(\text{red card}) = \frac{26}{52} = 0.5
\]

\[
P(\text{face card}) = \frac{12}{52} = \frac{3}{13} \approx 0.231
\]

\[
P(\text{queen}) = \frac{4}{52} = \frac{1}{13} \approx 0.077
\]

### Multiple Independent Observations

Let's return to coin flipping to illustrate situations where we have an event consisting of multiple independent observations. For example, the probability of throwing two consecutive heads is:

\[
P(\text{HH}) = \frac{1}{4} = 0.25
\]

Because there is one relevant outcome in the sample space of four possible events \(\Omega = \{\text{HH, HT, TH, TT}\}\).

Likewise, the probability of throwing three consecutive heads is:

\[
P(\text{HHH}) = \frac{1}{8} = 0.125
\]

Because there is one relevant outcome in the sample space of eight possible events \(\Omega = \{\text{HHH, HHT, HTH, HTT, THH, THT, TTH, TTT}\}\).

As a final example, the probability of throwing exactly two heads in three tosses is:

\[
P(\text{HHT, HTH, THH}) = \frac{3}{8} = 0.375
\]

### Conjunction Probabilities

To combine probabilities, we multiply them. So the probability of throwing two consecutive heads, for example, is the product of probabilities we've already calculated:

\[
P(\text{H}_1\text{H}_2) = P(\text{H}_1) \times P(\text{H}_2) = \frac{1}{2} \times \frac{1}{2} = \frac{1}{4} = 0.25
\]

### Combinatorics

Combinatorics is a field of mathematics devoted to counting that can help us to study probabilities. We can use **factorials** \(\text{e.g., } 3! = 3 \times 2 \times 1 = 6\), which function primarily in combinatorics, to calculate probabilities instead of painstakingly determining all of the number of the sample space \(\Omega\) and counting outcomes within it.

The left-hand side of this equation is read "\(n\) choose \(k\)" and is most quickly understood in this example. If we have three coin flips, \(n = 3\), and if we're interested in the number of ways to get two heads from two flips, for that \(k = 2\). We would read the left-hand side of this equation as "\(3\) choose \(2\)", and calculate it as:

\[
\binom{n}{k} = \frac{n!}{k!(n-k)!} \rightarrow \binom{3}{2} = \frac{3!}{2!(3-2)!} = \frac{3 \times 2 \times 1}{(2 \times 1)(1)} = \frac{6}{2} = 3
\]

This provides us with the numerator for our probability equation. For this calculation, we're interested in determining the number of ways we can achieve exactly two heads out of three flips. Therefore:

\[
P(\text{2 heads}) = \frac{\# \text{ of outcomes of event}}{\# \text{ of outcomes in } \Omega} = \frac{3}{8} = 0.375
\]

In the case of coin-flipping (or any binary process with equally probable outcomes), the denominator can be calculated with \(2^n\) where \(n\) equals the number of coin flips, so:

\[
\frac{\# \text{ of outcomes of event}}{\# \text{ of outcomes in } \Omega} = \frac{\binom{n}{k}}{2^n} = \frac{\binom{3}{2}}{2^3} = \frac{3}{8} = 0.375
\]

```python
from math import factorial
```

```python
def coinflip_prob(n, k):
    n_choose_k = factorial(n)/(factorial(k)*factorial(n-k))
    return n_choose_k/2**n
```

```python
coinflip_prob(5, 3)
```
