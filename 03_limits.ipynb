```python
def my_fxn(x):
    my_y = (x**2-1)/(x-1)
    return my_y
```

$$\lim_{x \to 0} \frac{\sin x}{x}$$

```python
def sin_fxn(x):
    y = (np.sin(x))/x
    return y
    
```

```python
# Uncommenting the following line results in a 'division by zero' error:
# y = sin_fxn(0)
```

```python
sin_fxn(0.1)
```

```python
sin_fxn(0.001)
```

```python
sin_fxn(-0.1)
```

```python
sin_fxn(-0.001)
```

```python
y = sin_fxn(x)
```

```python
plt.plot(x,y)
plt.axvline(x = 0, color = "lightgray")
plt.axhline(y = 0, color = "lightgray")

plt.xlim(-10,10)
plt.ylim(-1,2)

plt.axvline(x = 0, linestyle = "--")
plt.axhline(y = 1,  linestyle = "--")
```

$$\lim_{x \to \infty} \frac{25}{x}$$

```python
def inf_fxn(x):
    y = 25/x
    return y
```

```python
inf_fxn(1e3)
```

```python
inf_fxn(1e6)
```

```python
y = inf_fxn(x)
```

```python
plt.plot(x,y)
plt.axvline(x = 0, color = "lightgray")
plt.axhline(y = 0, color = "lightgray")

plt.xlim(-10,10)
plt.ylim(-300,300)

# plt.axvline(x = 0, linestyle = "--")
# plt.axhline(y = 1,  linestyle = "--")
```

```python
left_x = x[x<0]
right_x = x[x>0]
```

```python
left_y = inf_fxn(left_x)
right_y = inf_fxn(right_x)
```

```python
plt.plot(left_x, left_y, c='C0')
plt.plot(right_x, right_y, c='C0')
plt.axvline(x = 0, color = "lightgray")
plt.axhline(y = 0, color = "lightgray")

plt.xlim(-10,10)
plt.ylim(-300,300)

# plt.axvline(x = 0, linestyle = "--")
# plt.axhline(y = 1,  linestyle = "--")
```
